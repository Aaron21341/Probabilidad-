```python
from scipy.stats import norm

# Datos
media = 70
desviacion = 4
n_estudiantes = 500

# Probabilidad entre 68 y 72 kg
probabilidad = norm.cdf(72, media, desviacion) - norm.cdf(68, media, desviacion)

# Número de estudiantes
estudiantes = probabilidad * n_estudiantes

print(f"Probabilidad: {probabilidad:.4f}")
print(f"Porcentaje: {probabilidad*100:.2f}%")
print(f"Estudiantes: {round(estudiantes)}")
```
