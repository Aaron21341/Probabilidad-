```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import binom

# Parámetros
n = 10
p = 0.10

# Valores de x
valores_x = np.arange(0, n + 1)

# Probabilidades
probabilidades = binom.pmf(valores_x, n, p)

# Tabla de resultados
print("Valor de x\tProbabilidad")

for x, prob in zip(valores_x, probabilidades):
    print(f"{x}\t\t{prob:.10f}")

# Gráfica
plt.figure(figsize=(8,5))
plt.bar(valores_x, probabilidades, color='skyblue')
plt.title('Distribución Binomial n=10, p=0.10')
plt.xlabel('Valor de X')
plt.ylabel('Probabilidad')
plt.xticks(valores_x)
plt.grid(True, alpha=0.3)
plt.show()
```
