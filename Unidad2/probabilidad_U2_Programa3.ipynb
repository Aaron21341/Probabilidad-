```python
import math as mt
from itertools import permutations

n = 10  # asistentes
r = 4   # preguntas

Pr = mt.factorial(n) / mt.factorial(n-r)
print("Existen", Pr, "posibles permutaciones sin repetición")

# Si quieres ver las permutaciones explícitas:
asistentes = ["A1","A2","A3","A4","A5","A6","A7","A8","A9","A10"]
permutaciones_asistentes = list(permutations(asistentes, r))

for p in permutaciones_asistentes[:10]:  # muestra solo las primeras 10
    print(p)
```

```python
import math as mt
from itertools import product

n = 10  # asistentes
r = 4   # preguntas

Pr_rep = n ** r
print("Existen", Pr_rep, "posibles permutaciones con repetición")

# Si quieres ver ejemplos:
asistentes = ["A1","A2","A3","A4","A5","A6","A7","A8","A9","A10"]
permutaciones_rep = list(product(asistentes, repeat=r))

for p in permutaciones_rep[:10]:  # muestra solo las primeras 10
    print(p)
```
