```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

from google.colab import files
uploaded = files.upload()

autos = pd.read_csv("autos2.csv")
autos.head()
```

```python
autos['price'] = pd.to_numeric(autos['price'], errors='coerce')

precio_min = autos['price'].min()
precio_max = autos['price'].max()
precio_moda = autos['price'].mode()[0]
precio_media = autos['price'].mean()
precio_mediana = autos['price'].median()
precio_std = autos['price'].std()

print("Precio mínimo:", precio_min)
print("Precio máximo:", precio_max)
print("Moda de precios:", precio_moda)
print("Media de precios:", precio_media)
print("Mediana de precios:", precio_mediana)
print("Desviación estándar:", precio_std)
```
